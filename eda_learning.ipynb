```python
outlier_summary = {}

numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    outlier_summary[col] = len(outliers)

print(outlier_summary)
```

```python
plt.figure(figsize=(15,8))
sns.boxplot(data=df[numeric_cols])
plt.xticks(rotation=90)
plt.show()
```

```python
df_capped = df.copy()

for col in numeric_cols:
    Q1 = df_capped[col].quantile(0.25)
    Q3 = df_capped[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    df_capped[col] = df_capped[col].clip(lower_bound, upper_bound)
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.boxplot(df['CO(GT)'])
plt.title("Before Capping")

plt.subplot(1,2,2)
sns.boxplot(df_capped['CO(GT)'])
plt.title("After Capping")

plt.show()
```

```python
df = df_capped.copy()
```

```python
df.isnull().sum()
```

```python
df[numeric_cols] = df[numeric_cols].ffill()
```

```python
df[numeric_cols] = df[numeric_cols].ffill()
```

```python
df.isnull().sum()
```

```python
for col in numeric_cols:
    df[col + '_lag1'] = df[col].shift(1)
```

```python
df = df.dropna()
```

```python
train_size = int(len(df) * 0.8)

train = df.iloc[:train_size]
test = df.iloc[train_size:]
```

```python
from sklearn.metrics import mean_squared_error
results = {}

for col in numeric_cols:
    
    y_true = test[col]
    y_pred = test[col + '_lag1']
    
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    
    results[col] = {"MSE": mse, "RMSE": rmse}

print(results)
```

```python
plt.figure(figsize=(12,6))
plt.plot(test['CO(GT)'], label='Actual')
plt.plot(test['CO(GT)_lag1'], label='Naive Forecast')
plt.title('CO(GT) - Forecast vs Actual')
plt.xlabel('Time')
plt.ylabel('CO(GT)')
plt.legend()
plt.show()
```

```python
for col in numeric_cols:
    
    plt.figure(figsize=(10,5))
    
    plt.plot(test[col], label='Actual')
    plt.plot(test[col + '_lag1'], label='Naive Forecast')
    
    plt.title(f'{col} - Forecast vs Actual')
    plt.xlabel('Time')
    plt.ylabel(col)
    plt.legend()
    
    plt.show()
```

```python
window = 24

for col in numeric_cols:
    df[col + '_rolling_mean'] = df[col].rolling(window).mean()
    df[col + '_rolling_std'] = df[col].rolling(window).std()
```

```python
for col in numeric_cols:
    
    plt.figure(figsize=(10,5))
    
    plt.plot(df[col], alpha=0.4, label='Original')
    plt.plot(df[col + '_rolling_mean'], color='red', label='24h Rolling Mean')
    
    plt.title(f'{col} - Rolling Mean Trend')
    plt.xlabel('Time')
    plt.ylabel(col)
    plt.legend()
    
    plt.show()
```

```python
from statsmodels.graphics.tsaplots import plot_acf

for col in numeric_cols:
    
    plt.figure(figsize=(8,4))
    plot_acf(df[col], lags=50)
    plt.title(f'{col} - Autocorrelation')
    plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm')
plt.title("Pollutant Correlation Heatmap")
plt.show()
```

```python
window = 24

for col in numeric_cols:
    df[col + '_ma_forecast'] = df[col].rolling(window).mean().shift(1)
```

```python
df=df.dropna()
```

```python
train_size = int(len(df) * 0.8)

train = df.iloc[:train_size]
test = df.iloc[train_size:]
```

```python
ma_results = {}

for col in numeric_cols:
    
    y_true = test[col]
    y_pred = test[col + '_ma_forecast']
    
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    
    ma_results[col] = {"MSE": mse, "RMSE": rmse}

print(ma_results)
```

```python
comparison = {}

for col in numeric_cols:
    comparison[col] = {
        "Naive_RMSE": results[col]["RMSE"],
        "MA_RMSE": ma_results[col]["RMSE"]
    }

print(comparison)
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose
for col in numeric_cols:
    
    decomposition = seasonal_decompose(df[col], model='additive', period=24)
    
    fig = decomposition.plot()
    fig.set_size_inches(12,8)
    fig.suptitle(f"{col} - Seasonal Decomposition", fontsize=14)
    
    plt.tight_layout()
    plt.show()
```

```python
comparison_df = pd.DataFrame({
    "Naive_RMSE": {col: results[col]["RMSE"] for col in numeric_cols},
    "MA_RMSE": {col: ma_results[col]["RMSE"] for col in numeric_cols}
})

comparison_df
```

```python
comparison_df.plot(kind='bar', figsize=(12,6))

plt.title("Naive vs Moving Average RMSE Comparison")
plt.ylabel("RMSE")
plt.xticks(rotation=90)
plt.legend()
plt.show()
```

```python
for col in numeric_cols:
    df[col + '_lag1'] = df[col].shift(1)
    df[col + '_lag2'] = df[col].shift(2)
    df[col + '_lag3'] = df[col].shift(3)

df = df.dropna()
```

```python
train_size = int(len(df) * 0.8)

train = df.iloc[:train_size]
test = df.iloc[train_size:]
```

```python
from sklearn.linear_model import LinearRegression

lr_results = {}

for col in numeric_cols:
    
    X_train = train[[col + '_lag1', col + '_lag2', col + '_lag3']]
    y_train = train[col]
    
    X_test = test[[col + '_lag1', col + '_lag2', col + '_lag3']]
    y_test = test[col]
    
    model = LinearRegression()
    model.fit(X_train, y_train)
    
    y_pred = model.predict(X_test)
    
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    
    lr_results[col] = {"RMSE": rmse}

print(lr_results)
```

```python
final_comparison = pd.DataFrame({
    "Naive_RMSE": {col: results[col]["RMSE"] for col in numeric_cols},
    "MA_RMSE": {col: ma_results[col]["RMSE"] for col in numeric_cols},
    "LR_RMSE": {col: lr_results[col]["RMSE"] for col in numeric_cols}
})

final_comparison
```

```python
final_comparison.plot(kind='bar', figsize=(14,6))

plt.title("Model Comparison (RMSE)")
plt.ylabel("RMSE")
plt.xticks(rotation=90)
plt.legend()
plt.show()
```
